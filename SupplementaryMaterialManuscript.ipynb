<h1 style="text-align: center; color: red; font-weight: bold;">
  Machine Learning and Regression Models for Stability Assessment of Cantilever Retaining Walls-Supplementary Material Manuscript
</h1>


This file includes auxiliary information absent from the article titled "Machine Learning and Regression Models for Stability Assessment of Cantilever Retaining Walls," such as additional tables, figures, extensive formulas, and comprehensive result lists, which are essential for a complete understanding of the study, as well as for transparency and repeatability.

Three different datasets are analyzed to examine the impact of varying wall dimensions on failure scenarios. The first dataset, henceforth referred to as the *regular* set (E1), is taken from previous studies [1]. In this dataset, each design parameter has four levels, given by ($X_1$ : 0.25H, 0.50H, 0.75H, 1.0H), ($X_2$ : 0.15$X_1$, 0.30$X_1$, 0.45$X_1$, 0.60$X_1$), ($X_3$ : 0.06H, 0.09H, 0.12H, 0.15H), ($X_4$ : 0, 0.01, 0.02, 0.03), and \($\phi$ : 20$^{\circ}$, 27$^{\circ}$, 34$^{\circ}$, 41$^{\circ}$). The combinations of these design parameters correspond to ($4^5$  = 1024) different CRW designs. The second dataset (henceforth, *random*, or E2), comprising 1,251 different CRW designs, is generated randomly to ensure that it differs from the 1,024 wall combinations in the regular dataset while remaining within the minimum and maximum limits of the design parameters. The third dataset (henceforth, *mixed*, or E3) is obtained by combining E1 and E2. For every member of these three datasets, the safety factors ($F_s$), ($F_o$), and ($F_{ss}$) are computed using the methods described earlier.

<h4 style="color: red; font-weight: bold; text-align: left;">
  1. Performance Evaluation Metrics of Machine Learning Models
</h4>

The scalar model performance metrics used in the present study are listed in Table 2. They are the maximum error (MaxE), mean absolute error (MAE), mean absolute percentage error (MAPE), mean squared logarithmic error (MSLE), median absolute error (MedAE), Tweedie mean squared error (TMSE), Tweedie mean Poisson deviance (TMPE), Tweedie mean gamma deviance (TMGE), $D^2$ Tweedie score ($D^2$T), and $D^2$ pinball score ($D^2$P) [2, 3].

  <strong>Table 2. Evaluation metrics for regression models.</strong>
</div>


<table>

      <td><strong>MaxE</strong></td>
      <td>

        $\displaystyle \max_{i} \left|y_{a(i)}-y_{p(i)}\right|$
      </td>
      <td><strong>MPD</strong></td>
      <td>
        Mean Poisson Deviance (Tweedie with $p=1$)<br>
        $\displaystyle \frac{1}{n}\sum_{i=1}^{n} 2\left( y_{a(i)}\ln\left( \frac{y_{a(i)}}{y_{p(i)}} \right)+ y_{p(i)}-y_{a(i)}\right)$
      </td>
    </tr>
    <tr>
      <td><strong>MAE</strong></td>
      <td>
        Mean Absolute Error<br>
        $\displaystyle \frac{1}{n}\sum_{i=1}^{n} \left|y_{a(i)}-y_{p(i)}\right|$
      </td>
      <td><strong>MGD</strong></td>
      <td>
        Mean Gamma Deviance (Tweedie with $p=2$)<br>
        $\displaystyle \frac{1}{n}\sum_{i=1}^{n} 2\left( \ln\left( \frac{y_{p(i)}}{y_{a(i)}} \right)+ \frac{y_{a(i)}}{y_{p(i)}}-1\right)$
      </td>
    </tr>
    <tr>
      <td><strong>MAPE</strong></td>
      <td>
        Mean Absolute Percentage Error<br>
        $\displaystyle \frac{100}{n}\sum_{i=1}^{n} \frac{\left|y_{a(i)}-y_{p(i)}\right|}{y_{a(i)}}$
      </td>
      <td><strong>MTD</strong></td>
      <td>
        Mean Tweedie Deviance (Tweedie with $p=1.5$)<br>
        $\displaystyle \frac{1}{n}\sum_{i=1}^{n} 2\left( \frac{y_{a(i)}^{2-p}}{(1-p)(2-p)} - \frac{y_{a(i)}y_{p(i)}^{1-p}}{1-p} + \frac{y_{p(i)}^{2-p}}{2-p} \right)$
      </td>
    </tr>
    <tr>
      <td><strong>MSLE</strong></td>
      <td>
        Mean Squared Logarithmic Error<br>
        $\displaystyle \frac{1}{n}\sum_{i=1}^{n}\left[\ln(1+y_{a(i)})-\ln(1+y_{p(i)}) \right]^{2}$
      </td>
      <td><strong>D<sup>2</sup></strong></td>
      <td>
        Coefficient of determination for Tweedie deviance<br>
        $\displaystyle D^{2} = 1-\frac{D(y_{a}, y_{p})}{D(y_{a}, \bar{y}_{a})}$ where $\bar{y}_{a}$ is the mean of $y_{a}$
      </td>
    </tr>
    <tr>
      <td><strong>MedAE</strong></td>
      <td>
        Median Absolute Error<br>
        $\displaystyle \text{median}\left( \left|y_{a(1)}-y_{p(1)}\right|,\ldots, \left|y_{a(n)}-y_{p(n)}\right| \right)$
      </td>
      <td><strong>MPL</strong></td>
      <td>
        Mean Pinball Loss (quantile loss) with parameter $\alpha \in (0,1)$<br>
        $\displaystyle L_{\alpha} = \frac{1}{n}\sum_{i=1}^{n}\begin{cases}
          \alpha (y_{a(i)}-y_{p(i)}), & y_{a(i)} \geq y_{p(i)} \\
          (1-\alpha)(y_{p(i)}-y_{a(i)}), & y_{a(i)} < y_{p(i)}
        \end{cases}$
      </td>
    </tr>
  </tbody>
</table>

<p><em>* $y_{p(i)}$: Predicted value for $i$-th sample, $y_{a(i)}$: Actual value for $i$-th sample, $n$: number of samples, $\alpha=0.5$ for MPL.</em></p>

<h4 style="color: red; font-weight: bold; text-align: left;">
  5. Model input-output correlations
</h4> 

The resulting correlation coefficients and factors of safety are illustrated in Figure 1. Results reveal that $X_{1}/H$ (the overall aspect ratio of the CRW) and $X_3/H$ (the base thickness ratio) are, respectively, the most and least significant elements influencing stability.  Furthermore, it was noted that 48% of the $F_{s}$ values among the 2275 designs exceeded 1.1, 45% fell below 0.9, and 7% ranged between 0.9 and 1.1.  For $F_{o}$, the respective percentages were 76\%, 19\%, and 5\%.  Ultimately, for $F_{ss}$, the ratios were 71%, 8%, and 21%.  Upon analyzing these hazardous designs (i.e., $F_x\geq 1.10$), it was noted that the predominant failure mechanism is sliding.  In all instances of failure, the mean values of $X_1/H$, $X_2/X_1$, $X_b/X_1$, and $X_3/H$ were 0.70, 0.35, 0.50, and 0.10, respectively.
