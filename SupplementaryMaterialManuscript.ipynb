  Machine Learning and Regression Models for Stability Assessment of Cantilever Retaining Walls-Supplementary Material Manuscript
</h1>


This file includes auxiliary information absent from the article titled "Machine Learning and Regression Models for Stability Assessment of Cantilever Retaining Walls," such as additional tables, figures, extensive formulas, and comprehensive result lists, which are essential for a complete understanding of the study, as well as for transparency and repeatability.

Three different datasets are analyzed to examine the impact of varying wall dimensions on failure scenarios. The first dataset, henceforth referred to as the *regular* set (E1), is taken from previous studies [1]. In this dataset, each design parameter has four levels, given by ($X_1$ : 0.25H, 0.50H, 0.75H, 1.0H), ($X_2$ : 0.15$X_1$, 0.30$X_1$, 0.45$X_1$, 0.60$X_1$), ($X_3$ : 0.06H, 0.09H, 0.12H, 0.15H), ($X_4$ : 0, 0.01, 0.02, 0.03), and \($\phi$ : 20$^{\circ}$, 27$^{\circ}$, 34$^{\circ}$, 41$^{\circ}$). The combinations of these design parameters correspond to ($4^5$  = 1024) different CRW designs. The second dataset (henceforth, *random*, or E2), comprising 1,251 different CRW designs, is generated randomly to ensure that it differs from the 1,024 wall combinations in the regular dataset while remaining within the minimum and maximum limits of the design parameters. The third dataset (henceforth, *mixed*, or E3) is obtained by combining E1 and E2. For every member of these three datasets, the safety factors ($F_s$), ($F_o$), and ($F_{ss}$) are computed using the methods described earlier.

<h4 style="color: red; font-weight: bold; text-align: left;">
  1. Performance Evaluation Metrics of Machine Learning Models
</h4>

The scalar model performance metrics used in the present study are listed in Table 2. They are the maximum error (MaxE), mean absolute error (MAE), mean absolute percentage error (MAPE), mean squared logarithmic error (MSLE), median absolute error (MedAE), mean Poisson deviance (MPD), mean gamma deviance (MGD), mean Tweedie deviance (MTD), coefficient of determination for the Tweedie deviance ($D^2$), and mean pinball loss (MPL) [2, 3].

<p><em>* $y_{p(i)}$: Predicted value for $i$-th sample, $y_{a(i)}$: Actual value for $i$-th sample, $n$: number of samples, $\alpha=0.5$ for MPL.</em></p>


**Rationale for the choice of metrics.** These metrics have different sensitivities. MaxE, MAE and MedAE measure absolute error and are dominated by large targets (e.g., $F_o$ ranges from about 0.2 to 13), whereas MAPE and MSLE measure relative error and emphasize the small, design-critical values near $F_x=1$. The Poisson, gamma and Tweedie deviances are likelihood-based measures for positive, right-skewed targets such as factors of safety; they weight errors relative to the target, so errors at small factors of safety count more than in absolute-error metrics. $D^2$ generalizes $R^2$ to the Tweedie deviance, and the pinball loss at the median measures the quality of the central prediction. Because MSLE and the deviances require positive values, all metrics were evaluated on the absolute value of the predictions. Negative predictions occur only for the under-fitting LR, RR and BR models at small factors of safety; the absolute value can only reduce their apparent error, so the model ranking in Section 4.1 of the article is conservative.

<h4 style="color: red; font-weight: bold; text-align: left;">
  6. Descriptions of the Machine Learning Models
</h4>

**Bayesian Ridge Regression (BR).** BR [5] is a probabilistic form of ridge regression in which the coefficients are random variables. The target is modelled as $y \sim \mathcal{N}(\mathbf{x}^{T}\boldsymbol{\beta},\,\alpha^{-1})$ with a zero-mean Gaussian prior $\boldsymbol{\beta} \sim \mathcal{N}(\mathbf{0},\,\lambda^{-1}\mathbf{I})$, and the posterior follows from Bayes' rule,

$$
p(\boldsymbol{\beta}\,|\,\mathbf{X}, y) \propto p(y\,|\,\mathbf{X}, \boldsymbol{\beta},\alpha)\, p(\boldsymbol{\beta}\,|\,\lambda),
$$

where the precisions $\alpha$ and $\lambda$ have Gamma hyper-priors with parameters $(\alpha_1, \alpha_2)$ and $(\lambda_1, \lambda_2)$. These are estimated by maximizing the marginal likelihood, so the amount of regularization is learned from the data. BR thus accounts for parameter uncertainty, limits overfitting and gives predictive distributions rather than point predictions.

**k-Nearest Neighbors (kNN).** In kNN regression [6], the prediction for a query point is the average target value of its $k$ closest training samples. Proximity is measured with the Minkowski distance,

$$
D(\mathbf{x},\mathbf{z}) = \left( \sum_{i=1}^{m} \left| x_i - z_i \right|^{p} \right)^{1/p},
$$

where $m$ is the number of input features; the Euclidean distance ($p=2$) is used here. The `KNeighborsRegressor` of Scikit-Learn is employed, and $k$ is selected by the hyperparameter search (Section 3.2 of the article).

**Support Vector Regression (SVR).** SVR, the regression counterpart of the Support Vector Machine (SVM), models nonlinear relationships through a kernel function. The `SVR` class of Scikit-Learn has two main parameters: $C$, which balances data fit against model complexity, and $\epsilon$, the width of the insensitive tube within which errors are not penalized. Its training cost grows roughly quadratically with the number of samples [7], which must be considered for large synthetic datasets.
